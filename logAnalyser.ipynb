```python
d = {}

with open("sample.log", "r") as f:
    for line in f:                      
        parts = line.split()           

        if len(parts) >= 3:             
            level = parts[3].strip("[]")  

            if level in d:
                d[level] += 1
            else:
                d[level] = 1

print(d)
    
```
